**Fill missing values with mean**

```python
# Fill missing values with mean
df["Age"].fillna(df["Age"].mean(), inplace=True)

df["Monthly_Income"].fillna(df["Monthly_Income"].mean(), inplace=True)

df["Credit_Score"].fillna(df["Credit_Score"].mean(), inplace=True)

df["Loan_Amount"].fillna(df["Loan_Amount"].mean(), inplace=True)

df["Previous_Loans"].fillna(df["Previous_Loans"].mean(), inplace=True)
```

```python
print(df.isnull().sum())
```

**Fix Inconsistent Data**

```python
# Standardize Loan_Approved values
df["Loan_Approved"] = df["Loan_Approved"].replace({
    "YES": "Yes",
    "yes": "Yes",
    "NO": "No"
})
```

```python
print(df["Loan_Approved"].value_counts())
```

**Fix invalid ages**

```python
# Replace invalid age with median
median_age = df["Age"].median()

df.loc[(df["Age"] < 18) | (df["Age"] > 100), "Age"] = median_age
```

**Fix invalid credit score:**

```python
median_credit = df["Credit_Score"].median()

df.loc[(df["Credit_Score"] < 300) |
       (df["Credit_Score"] > 850),
       "Credit_Score"] = median_credit
```

**Remove duplicates:**

```python
df = df.drop_duplicates()
```

```python
print(df.duplicated().sum())
```

**Treat Noisy/Outlier Values**

***Using IQR method***

```python
numeric_cols = [
    "Age",
    "Monthly_Income",
    "Credit_Score",
    "Loan_Amount",
    "Previous_Loans"
]

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # Replace outliers with median
    median = df[col].median()

    df.loc[(df[col] < lower) | (df[col] > upper), col] = median
```

```python
# Fix negative Monthly Income
median_income = df["Monthly_Income"].median()

df.loc[df["Monthly_Income"] < 0,
       "Monthly_Income"] = median_income
```

```python
# Fix negative Loan Amount
median_loan = df["Loan_Amount"].median()

df.loc[df["Loan_Amount"] < 0,
       "Loan_Amount"] = median_loan
```

```python
# Fix negative Previous Loans
median_prev = df["Previous_Loans"].median()

df.loc[df["Previous_Loans"] < 0,
       "Previous_Loans"] = median_prev
```
